```python
import pandas as pd

# 1. Cargar el archivo Excel
file = './IncendiosForestales/estadisticasIncendiosForestales2015-2025_TJmHngW.xlsx'
print("Leyendo el archivo Excel (esto puede tardar unos segundos)...")
df = pd.read_excel(file)

# 2. Imprimir exploración (lo que ya tenías)
print("Lista de columnas para el Cubo OLAP:")
for col in df.columns:
    print(f"- {col}")

print("\nResumen de tipos de datos:")
print(df.info())

# 3. EXPORTAR A CSV (La nueva magia)
archivo_csv = './IncendiosForestales/incendios_forestales_2015_2025.csv'
print(f"\nConvirtiendo a CSV y guardando en: {archivo_csv} ...")

# index=False evita que Pandas agregue una columna extra con los números de fila
# encoding='utf-8-sig' asegura que las 'ñ' y los acentos (como en "Año" o "Región") se guarden perfecto
df.to_csv(archivo_csv, index=False, encoding='utf-8-sig')

print("¡Conversión a CSV completada con éxito!")
```

```python
import pandas as pd
df = pd.read_csv('/home/saul/Metodologia/IncendiosForestales/01_ExtraccionDeDatos/DatosConafor/raw/incendios_forestales_2015_2025.csv', dtype=str)
df_2025 = df[df['Año'] == '2025']

print("--- DIAGNÓSTICO DE COLUMNAS ---")
print(df_2025.columns.tolist()) # Ver si hay espacios ocultos

print("\n--- MUESTRA DE DATOS 2025 ---")
print(df_2025[['Fecha Inicio', 'Latitud', 'Longitud']].head())


# 1. Primero, veamos todos los nombres de estados que hay en 2025
# Esto nos dirá si dice "México", "Estado de México" o "MEXICO"
print("\n--- NOMBRES DE ESTADOS ENCONTRADOS EN 2025 ---")
print(df_2025['Estado'].unique())

# 2. Ahora filtramos específicamente por lo que contenga "Méx" 
# (así atrapamos cualquier variación con o sin acento)
df_solo_mexico = df_2025[df_2025['Estado'].str.contains('Méx|Mex', na=False, case=False)]

print(f"\n--- REVISIÓN DETALLADA DE MÉXICO (2025) ---")
print(f"Total de filas encontradas: {len(df_solo_mexico)}")

if not df_solo_mexico.empty:
    # Ver los primeros 10 para notar patrones en Fecha o Coordenadas
    print("\nPrimeros 10 registros:")
    print(df_solo_mexico[['Estado', 'Fecha Inicio', 'Latitud', 'Longitud']].head(10))
    
    # Prueba de fuego: ¿Qué pasa si intentamos convertir las fechas de estos 1308 registros?
    fechas_convertidas = pd.to_datetime(df_solo_mexico['Fecha Inicio'], errors='coerce')
    print(f"\nResultado de conversión de fecha:")
    print(f"Fechas válidas: {fechas_convertidas.notna().sum()}")
    print(f"Fechas que fallaron (NaT): {fechas_convertidas.isna().sum()}")
else:
    print("¡Error Crítico! No se encontró ningún registro que coincida con 'México' en 2025.")
```

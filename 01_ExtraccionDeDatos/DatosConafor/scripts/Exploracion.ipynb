```python
import pandas as pd

# Cargar el archivo (asegúrate de que el nombre sea exacto)
file = './IncendiosForestales/incendios_forestales_2015_2025.csv'
df = pd.read_csv(file)

# Imprimir todas las columnas una por una
print("Lista de columnas para el Cubo OLAP:")
for col in df.columns:
    print(f"- {col}")

# Ver un resumen de los datos
print("\nResumen de tipos de datos:")
print(df.info())
```

```python
import pandas as pd

# 1. Cargar el archivo Excel
file = './IncendiosForestales/estadisticasIncendiosForestales2015-2025_TJmHngW.xlsx'
print("Leyendo el archivo Excel (esto puede tardar unos segundos)...")
df = pd.read_excel(file)

# 2. Imprimir exploración (lo que ya tenías)
print("Lista de columnas para el Cubo OLAP:")
for col in df.columns:
    print(f"- {col}")

print("\nResumen de tipos de datos:")
print(df.info())

# 3. EXPORTAR A CSV (La nueva magia)
archivo_csv = './IncendiosForestales/incendios_forestales_2015_2025.csv'
print(f"\nConvirtiendo a CSV y guardando en: {archivo_csv} ...")

# index=False evita que Pandas agregue una columna extra con los números de fila
# encoding='utf-8-sig' asegura que las 'ñ' y los acentos (como en "Año" o "Región") se guarden perfecto
df.to_csv(archivo_csv, index=False, encoding='utf-8-sig')

print("¡Conversión a CSV completada con éxito!")
```
